### 🧠 Exercise: Dictionary Practice

```python
# TODO: Create a dictionary of 3 students with names as keys and scores as values
students = {
    "Azeez": 75,
    "Lanre": 82,
    "Kemi": 78
}

# TODO: Find the student with the highest score
highest_student = max(students, key=students.get)
print(f"Student with highest score: {highest_student}")

# TODO: Add a new key 'grade' for each student based on score
for name, score in students.items():
    if score >= 80:
        students[name] = (score, 'A')
    elif score >= 70:
        students[name] = (score, 'B')
    elif score >= 60:
        students[name] = (score, 'C')
    else:
        students[name] = (score, 'D')
        
print("Students with Grades:")
for name, (score, grade) in students.items():
    print(f"{name}: Score = {score}, Grade = {grade}")
```

### 🧠 Exercise: Algorithm Practice

```python
# TODO: Implement binary search on a sorted list
def binary_search(arr, target):
        low = 0 # Starting index
        high = len(arr) - 1 # Ending index
        
        while low <= high:
            mid = (low + high) // 2 # Middle index
            if arr[mid] == target: 
                return mid
            elif arr[mid] < target:
                low = mid + 1 # Move to the right half
            else:
                high = mid - 1 # Move to the left half
        return -1 # Target not found
    
sorted_list = [1, 3, 5, 7, 9, 11, 13, 15]
target = 7
result = binary_search(sorted_list, target)
if result != -1:
    print(f"Element found at index: {result}")
else:
    print("Element not found in the list.")
    
# TODO: Implement a function to find the factorial of a number (iteratively or recursively)
def recursive_factorial(n):
    """
    Calculate the factorial of a number n recursively.
    """
    if n == 0 or n == 1:
        return 1
    else:
        return n * recursive_factorial(n - 1)
    
def iterative_factorial(n):
    """
    Calculate the factorial of a number n iteratively.
    """
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

number = 5
print(f"Recursive Factorial of {number}: {recursive_factorial(number)}")
print(f"Iterative Factorial of {number}: {iterative_factorial(number)}")

# TODO: Write a function to find the largest number in a list without using max()
def find_largest_number(numbers: list) -> int:
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest

number_list = [34, 67, 23, 89, 12, 90, 45]
largest_number = find_largest_number(number_list)
print(f"Largest Number in the list: {largest_number}")
```
