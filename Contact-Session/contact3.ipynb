```python
# Write a function to check if a number is prime
def is_prime(num):
    for i in range(2, num):
        if num%i == 0:
            return "False"
    return "True"


print(is_prime(5))
```

```python
# Create a Student class with attribute name, age, grade. Add a method get_details() that returns a formatted string
class Student(): 
    def __init__(self, name, age, grade): 
        self.name = name 
        self.age = age 
        self.grade = grade
        
    def get_details(self): 
        return f"Name: {self.name}\nAge: {self.age}\nGrade: {grade}"
    
    
std1 = Student("John", 15, "10th")
print(std1.get_details())
```
